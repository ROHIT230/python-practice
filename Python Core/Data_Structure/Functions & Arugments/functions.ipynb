Anonymous Functions


In Python, an anonymous function means that a function is without a name. As we already know the def keyword is used to define the normal functions and the lambda keyword is used to create anonymous functions.

Python Lambda Functions

Python Lambda Functions


Lambda functions are small anonymous functions, meaning they do not have a defined name. These are small, short-lived functions used to pass simple logic to another function.

Contain only one expression.
Result of that expression is returned automatically (no return keyword needed).

```python
a= "Geeks for Geeks"

upper = lambda x: x.upper()

print(upper(a))
```

Use Cases of Lambda Functions

Let's see some of the practical uses of the Python lambda function.

1. Using with Condition Checking

A lambda function can use conditional expressions (if-else) to return different results based on a condition.

```python
check = lambda x: "positve" if x>2 else "Negatve " if x<2 else "Zero"

print(check(1))
```

2. Using with List Comprehension

Lambda functions can be combined with list comprehensions to apply the same operation to multiple values in a compact way.

```python
fucn =[ lambda arg = x : arg *10 for x in range(1,5)]

for i in fucn:
    print(i)
```

```python
fucn =[ lambda arg = x : arg *10 for x in range(1,5)]

for i in fucn:
    print(i())
```

```python
fucn =[ lambda arg = x : arg *10 for x in range(1,5)]

print([f() for f in fucn])
```

Important note

3. Using for Returning Multiple Results

Although a lambda can contain only one expression, it can still return multiple results by combining them into a tuple.

```python
cal = lambda x, y : (x+y,x*y)
print(cal(1,2))
```

 Using with filter()

filter() function uses a lambda expression to select elements from a list that satisfy a given condition, such as keeping only even 
numbers.

```python
c = [1, 2, 3, 4, 5, 6]

filterd_data= filter(lambda x: x%2 ==0,c)
print(list(filterd_data))
```

Using with map()

map() function applies a lambda expression to each element of a list and returns a new list with the transformed values.

```python
a = [1, 2, 3, 4]

double= map(lambda x : x*2,a)


print(list(double))
```

. Using with reduce()

reduce() function repeatedly applies a lambda expression to elements of a list to combine them into a single result.

```python
from functools import reduce

a = [1, 2, 3, 4]

reduced_value = reduce(lambda x,y: x*y , a)

print(reduced_value)
```

Python is NOT purely call by value or call by reference

👉 It uses:

Call by Object Reference (or Call by Assignment)

🧠 What Does That Mean?

When you pass a variable to a function:

👉 You pass a reference to the object
👉 But behavior depends on mutable vs immutable

Case 1: Immutable Objects (Behaves like Call by Value)

```python
def func(x):
    x = x + 10

a = 5
func(a)

print(a)


# Why?

# x points to 5

# x = x + 10 creates new object (15)

# a still points to 5
```

Key Point

👉 Immutable → changes create new object → original unaffected

🔥 Case 2: Mutable Objects (Behaves like Call by Reference)

```python
def func(lst):
    lst.append(100)

nums = [1,2,3]
func(nums)

print(nums)
```

Interview-Level Explanation

If interviewer asks:

Is Python call by value or reference?

Answer:

Python uses call by object reference. The function receives a reference to the object, and behavior depends on whether the object is mutable or immutable.

```python
def func(lst):
    lst += [4]

nums = [1,2,3]
func(nums)

print(nums)
```

```python
# Q3 (Tricky)
def func(lst):
    lst = lst + [4]

nums = [1,2,3]
func(nums)

print(nums)
```

Interview Answer (Short & Perfect)

If asked:

Difference between lst += [4] and lst = lst + [4]?

Say:

+= modifies the list in-place, while lst = lst + [4] creates a new list and reassigns it, so the original list remains unchanged."

Key Difference
Operation	Behavior
+=	modifies existing object
   "+ creates new object
