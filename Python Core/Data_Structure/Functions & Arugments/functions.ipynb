```python
# default argument

def myfunc(x,y=50):
    print(x)
    print(y)


myfunc(10)
```

```python
# keyword function argument

def myfunc(fname,lname):
    print(lname)
    print(fname)



myfunc(fname="rohit",lname="rana")
```

```python
# postional argument

def myfunc(fname,lname):
    print("Hi i am ",fname)
    print("my last name is ", lname)


myfunc("rohit","rana")
```

```python
#arbitrary arguments


def myfunc(*args,**kwargs):

    print(type(args))
    print(type(kwargs))

    for arg in args:
        print(arg)

    for key, value in kwargs.items():
        print(key,"--------->",value)


myfunc("rohit","rana",fname="rohit",lname="rana")
```

Function within Functions
A function defined inside another function is called an inner function (or nested function). It can access variables from the enclosing function’s scope and is often used to keep logic protected and organized.

```python
def func1():
    s="Hello World"
    def func2():
        print(s)

    func2()


func1 #### when we do not add () after function name ()----> it will return lie "<function __main__.func1()>"
func1()
```
