Using Lambda + map

Question:

Square all numbers in a list

Input: [1,2,3,4]
Output: [1,4,9,16]

```python
a= [1,2,3,4]
squaere_a = list(map(lambda x: x*2, a))

print(squaere_a)

```

Using filter

Question:
Get only even numbers

Input: [1,2,3,4,5,6]
Output: [2,4,6]

```python
b= [1,2,3,4,5,6]

even_no= list(filter(lambda x : x%2==0,b))

print(even_no)
```

Using reduce

Question:
Find product of all elements

Input: [1,2,3,4]
Output: 24

```python
from functools import reduce

c = [1,2,3,4]

reduced_c = reduce(lambda x, y: x*y,c)
print(reduced_c)
```

List Comprehension (Important)

Question:
Get squares of even numbers only

Input: [1,2,3,4,5]
Output: [4,16]

```python
a= [1,2,3,4,5]

b=[x**2 for x in a if x%2==0]
print(b)
```

Convert list of strings to uppercase
Input: ["python","java"]
Output: ["PYTHON","JAVA"]

```python
a= ["python","java"]

result = list(map(lambda x: x.upper(),a ))

print(result)
```

Dictionary Comprehension

Question:
Create dictionary of numbers and their squares

Input: [1,2,3]
Output: {1:1, 2:4, 3:9}

```python
a = [1,2,3]
b= {x:x*2 for x in a}

print(b)
```

Combine map + filter

Question:
Square only even numbers
Input: [1,2,3,4]
Output: [4,16]

```python
a= [1,2,3,4]

result = list(map(lambda x: x**2, filter(lambda x: x%2==0,a)))
print(result)
```

Flatten list using comprehension

Input: [[1,2],[3,4]]
Output: [1,2,3,4]

```python
a= [[1,2],[3,4]]

b=[y for x in a for y in x]
print(b)
```

```python
# Mutable Default Argument (MOST IMPORTANT)

def func(a, lst=[]):
    print(lst)
    lst.append(a)
    return lst

print(func(1))
print(func(2))
```

```python
# Best Practice
def func(a, lst=None):
    if lst is None:
        lst = []
    lst.append(a)
    return lst


print(func(1))

print(func(2))
# 👉 Always avoid mutable defaults
```

```python
# Reassignment vs Mutation
def func(lst):
    lst = [10,20]
    
nums = [1,2,3]
func(nums)

print(nums)

# Why?

# lst = [...] creates a new object

# It does NOT affect original list
```

```python
# Mutation Case
def func(lst):
    lst.append(100)

nums = [1,2,3]
func(nums)

print(nums)

# 💡 Why?

# Lists are mutable

# Same object is modified
```

```python
# Best Practice

# If you want to avoid modification:

def func(lst):
    lst = lst.copy()
```

```python
# kwargs Trick
def func(**kwargs):
    kwargs["age"] = 40

data = {"name": "Rohit", "age": 30}

func(**data)

print(data)
# ✅ Output
# {'name': 'Rohit', 'age': 30}
# 💡 Why?

# **data creates a copy

# Changes inside function don’t affect original dict
```
