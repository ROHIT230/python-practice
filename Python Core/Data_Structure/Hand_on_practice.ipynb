Challenge 6 – Most Frequent Element in List

Example:

Input: [1,2,2,3,3,3,4]
Output: 3

Hint:

Use:

Counter


What is most_common()?

most_common() returns elements sorted by frequency.

```python
from collections import Counter

def most_frequesnt_element(lst):
    a= Counter(lst)
    freq_element= a.most_common(1)[0][0]
    return freq_element



print(most_frequesnt_element([1,2,2,3,3,3,4]))
```

Challenge 7 – Flatten a Nested List

Example:

Input: [[1,2],[3,4],[5]]
Output: [1,2,3,4,5]

```python
def flat_nested_list(lst):
    b=[]
    for i in lst:
        b.extend(i)


    return b



print(flat_nested_list([[1,2],[3,4],[5]]))
```

Find Missing Number

You are given numbers from 1 to n, but one number is missing.

Example:

nums = [1,2,3,5]

Output

4

```python
def missing_number(lst):
    n =len(lst) + 1
    total =  n*(n+1)//2
    return total - sum(lst)


print(missing_number([1,2,3,4,5,7]))
```

First Non-Repeating Character

Example:

Input: "aabbcdde"
Output: c

```python
from collections import Counter

def check_non_rep_chr(word):
    count = Counter(word)

    for ch in word:
        if count[ch]==1:
            return ch
    
    
print(check_non_rep_chr("aabbcdde"))
```

Find Duplicate in List

Example

Input: [1,2,3,4,2,5]
Output: 2

```python
def duplicates(lst):
    seen =set()

    for num in lst:
        if num in seen:
            return num
        
        seen.add(num)

print(duplicates([1,2,3,3,3,4,2,5]))
```

Rotate List

Example

Input: [1,2,3,4,5]
Rotate by 2

Output:
[4,5,1,2,3]

```python
def rotate_list(lst,k ):
    k = k % len(lst)
    lst= lst[-k:]+ lst[:-k]
    return lst

rotate_list([1,2,3,4,5],6)
```

```python
from collections import Counter
def first_non_repeating_chr(word):
    count = Counter(word)
    for i in word:
        if count[i]==1:
            return i
        

first_non_repeating_chr("swiss")
```
