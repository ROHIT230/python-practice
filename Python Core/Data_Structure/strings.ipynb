String Immutability
Strings are immutable, which means that they cannot be changed after they are created. If we need to manipulate strings then we can use methods like concatenation, slicing or formatting to create new strings based on original.

```python
s = "geeksforGeeks"

s=  "G" + s[1:]
print(s)
```

Updating a String
As strings are immutable, “updates” create new strings using slicing or methods such as replace().

Example: This code fixes the first letter and replace a word.

```python
s = "hello geeks"
s1 = "H" + s[1:]                  
s2 = s.replace("geeks", "GeeksforGeeks")  
print(s1)
print(s2)
```

Challenge 1 – Remove Duplicates From String

Question

Write a Python function to remove duplicate characters from a string while keeping the order.

Example:

Input:  "programming"
Output: "progamin"

Try solving it.

Hint:
You can use set + loop.

```python
def remove_duplicates(a:str)->str:
    seen = set()
    word =""
    for i in a :
        if i not in seen:
            seen.add(i)
            word += i
    return word

print(remove_duplicates("programming"))
        

```

Challenge 3 – Reverse Words in a Sentence

Example:

Input:  "python is powerful"
Output: "powerful is python"

Expected approach:

split

reverse

join

```python
def reverse(word:str):
    word_list= word.split(" ")
    print(word_list)

    word_list.reverse()
    return " ".join(word_list)



print(reverse("python is powerful"))
```
