Iterators in Python

An iterator in Python is an object used to traverse through all the elements of a collection (like lists, tuples or dictionaries) one 
element at a time. It follows the iterator protocol, which involves two key methods:

__iter__(): Returns the iterator object itself.

__next__(): Returns the next value from the sequence. Raises StopIteration when the sequence ends.

Why do we need iterators?

Here are some key benefits:


Lazy Evaluation: Processes items only when needed, saving memory.

Generator Integration: Pairs well with generators and functional tools.

Stateful Traversal: Keeps track of where it left off.

Uniform Looping: Same for loop works for lists, strings and more.

Composable Logic: Easily build complex pipelines using tools like itertools.


```python
s="GFG"
it=iter(s)

print(next(it))
print(next(it))
```

Creating a Custom Iterator

```python
class EvenNo():
    def __init__(self):
        self.number= 2

    def __iter__(self):
        return self
    
    def __next__(self):
        n= self.number

        self.number+=2

        return n
    

ev= EvenNo()
it= iter(ev)
print(next(it))
print(next(it))
```

StopIteration Exception
StopIteration exception is integrated with Python’s iterator protocol. It signals that the iterator has no more items to return. Once this exception is raised, further calls to next() on the same iterator will continue raising StopIteration.

```python
class EvenNo():
    def __init__(self):
        self.number= 2

    def __iter__(self):
        return self
    
    def __next__(self):
       
        n= self.number

        self.number+=2

        return n
      

ev= EvenNo()

print(next(ev))
print(next(ev))


```

```python
li = [100, 200, 300]

it= iter(li)


while True:
    try:
        print(next(it))

    except StopIteration:
        print("End of iter")
        break
```

Practical Example 2: Custom Counter (Manual Iterator)

```python
class Counter:
    def __init__(self, max):
        self.max=max
        self.current= 0

    def __iter__(self):
        return self


    def __next__(self):
        if self.current< self.max:
            self.current +=1
            return self.current
        else:
            raise StopIteration
        



ob1= Counter(3)
for i in ob1:
    print(i)
```

Database Batch Processing
Problem:

Process data in chunks (common in backend systems)

```python
class Loader:
    def __init__(self,data,batch_size):
        self.data= data
        self.batch_size= batch_size
        self.index=0

    def __iter__(self):
        return self
    
    def __next__(self):
        if self.index>= len(self.data):
            raise StopIteration
        
        batch= self.data[self.index:self.index+self.batch_size]
        self.index+=self.batch_size
        return batch


data = [1,2,3,4,5,6,7]

dl= Loader(data,3)
for batch in dl:
    print(batch)
```
