What is a Generator?

👉 A generator is a function that:

Uses yield instead of return
Produces values one at a time
Maintains its state between calls

```python
def gen():
    yield 1
    yield 2
    yield 3

g=gen()

print(next(g))
```

💡 Key Concept

👉 yield:

Pauses function
Saves state
Resumes from same point

Difference
Feature	Normal Function	Generator

Execution	All at once	One by one

Memory	High	Low

Keyword	return	yield


```python
def even_no():
    n=0
    while True:
        yield n
        n+=2 
        

g= even_no()

for _ in range(5):

    print(next(g))
        
```

mportant Concept: Lazy Evaluation

👉 Generator does NOT compute everything at once

Example:

g = (x*x for x in range(1000000))

👉 No memory issue
👉 Values generated only when needed

🔥 Generator vs List
# List
lst = [x*x for x in range(1000000)]

# Generator
gen = (x*x for x in range(1000000))

👉 List → huge memory
👉 Generator → efficient

```python
g = (x for x in range(3))

print(list(g))
print(list(g))
```
